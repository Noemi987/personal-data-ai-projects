```python
#Creazione di una copia di lavoro
annunci_puliti = annunci.copy()

#Rimozione degli spazi superflui nei campi identificativi
annunci_puliti["ID_annuncio"] = annunci_puliti["ID_annuncio"].str.strip()
annunci_puliti["ID_ruolo"] = annunci_puliti["ID_ruolo"].str.strip()
annunci_puliti["Profilo"] = annunci_puliti["Profilo"].str.strip()

#Verifica della distribuzione dopo la pulizia
annunci_puliti[["ID_ruolo", "Profilo"]].value_counts()
```

```python
#Pulizia degli spazi negli altri campi testuali

annunci_puliti["Titolo_originale"] = annunci_puliti["Titolo_originale"].str.strip()
annunci_puliti["Azienda"] = annunci_puliti["Azienda"].str.strip()
annunci_puliti["Localita"] = annunci_puliti["Localita"].str.strip()
annunci_puliti["URL"] = annunci_puliti["URL"].str.strip()
annunci_puliti["Data_raccolta"] = annunci_puliti["Data_raccolta"].str.strip()
annunci_puliti["Esperienza_richiesta"] = annunci_puliti["Esperienza_richiesta"].str.strip()
annunci_puliti["Competenze"] = annunci_puliti["Competenze"].str.strip()
annunci_puliti["Modalita_lavorativa"] = annunci_puliti["Modalita_lavorativa"].str.strip()
annunci_puliti["Benefit"] = annunci_puliti["Benefit"].str.strip()
annunci_puliti["Crescita"] = annunci_puliti["Crescita"].str.strip()

#Verifica delle dimensioni
annunci_puliti.shape
```

```python
#Controlli finali sul dataset pulito
print("Valori nulli:", annunci_puliti.isna().sum().sum())
print("Celle vuote:", annunci_puliti.eq("").sum().sum())
print("ID duplicati:", annunci_puliti["ID_annuncio"].duplicated().sum())
print("URL duplicati:", annunci_puliti["URL"].duplicated().sum())
```

```python
#Esportazione del dataset pulito
annunci_puliti.to_csv(
    "../03_dati_puliti/annunci_puliti.csv",
    index=False,
    sep=",",
    encoding="utf-8-sig"
)

print("Esportazione completata!")
print("Dimensioni del dataset:", annunci_puliti.shape)

#to_csv() esporta il DataFrame in formato CSV
#../03_dati_puliti/ indica la cartella di destinazione
#index=False evita di aggiungere una colonna numerica inutile
#sep="," mantiene il separatore già utilizzato nel CSV originale
#encoding="utf-8-sig" preserva i caratteri accentati e facilita la lettura del file
```

```python
#Rilettura del CSV esportato

verifica_csv = pd.read_csv(
    "../03_dati_puliti/annunci_puliti.csv",
    sep=",",
    encoding="utf-8-sig"
)

#Controllo delle dimensioni e dell'integrità dei dati
print("Dimensioni file riletto:", verifica_csv.shape)
print("Dati identici alla copia pulita:", verifica_csv.equals(annunci_puliti))
```
